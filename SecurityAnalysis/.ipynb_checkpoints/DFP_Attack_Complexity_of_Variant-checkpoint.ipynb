```python
import numpy as np
import matplotlib.pyplot as plt
from math import log, exp, e


# Define function for final complexity
def compute_complexity_c(k, threshold, m):
    # Compute log(prob1) safely
    #print(f"The value of k is {k}, dimension is {m}, threshold = {threshold}")
    log_prob1 = m * log(k, e) + (m / 2) * (1 - k**2)  # Natural log of prob1
    classical_complexity = -log_prob1 / log(2)  # Convert to base 2
    quantum_complexity = classical_complexity / 2

    kappa = threshold / k
    #print(f"The value of kappa is {kappa}")
    log_prob2 = log(2, e) - (kappa**2) / 2  # Compute log(prob2) safely
    decryption_failure_complexity = -log_prob2 / log(2)  # Convert to base 2

    classical_complexity = classical_complexity + decryption_failure_complexity
    quantum_complexity = quantum_complexity + decryption_failure_complexity
    #print("\n")
    return classical_complexity

# Generate values for k
k_values = np.arange(1, 1.5, 0.05)

# Variant-512
rank = 2 #2,3,4
n = 256
#logDFP = -139
logDFP = -146.1
m = 2*rank*n
threshold = sqrt(2/log(e,2) - 2*logDFP/log(e,2))
final_complexities1_c = [compute_complexity_c(k, threshold, m) for k in k_values]

# Variant-768
rank = 3 #2,3,4
n = 256
#logDFP = -164
logDFP = -177.8
m = 2*rank*n
threshold = sqrt(2/log(e,2) - 2*logDFP/log(e,2))
final_complexities2_c = [compute_complexity_c(k, threshold, m) for k in k_values]

# Variant-1024
rank = 4 #2,3,4
n = 256
#logDFP = -174
logDFP = -167.6
m = 2*rank*n
threshold = sqrt(2/log(e,2) - 2*logDFP/log(e,2))
final_complexities3_c = [compute_complexity_c(k, threshold, m) for k in k_values]

```

```python
for i in range(len(k_values)):
    print(format(k_values[i], '.2f'), end = ", \t")
    print(format(final_complexities1_c[i],'.2f'), end = ", \t")
    print(format(final_complexities2_c[i],'.2f'), end = ", \t")
    print(format(final_complexities3_c[i],'.2f'))
```

# Quantum Complexity

```python
import numpy as np
import matplotlib.pyplot as plt
from math import log, exp, e


# Define function for final complexity
def compute_complexity_q(k, threshold, m):
    # Compute log(prob1) safely
    #print(f"The value of k is {k}, dimension is {m}, threshold = {threshold}")
    log_prob1 = m * log(k, e) + (m / 2) * (1 - k**2)  # Natural log of prob1
    classical_complexity = -log_prob1 / log(2)  # Convert to base 2
    quantum_complexity = classical_complexity / 2.0
    

    kappa = threshold / k
    #print(f"The value of kappa is {kappa}")
    log_prob2 = log(2, e) - (kappa**2) / 2  # Compute log(prob2) safely
    decryption_failure_complexity = -log_prob2 / log(2)  # Convert to base 2

    classical_complexity = classical_complexity + decryption_failure_complexity
    quantum_complexity = quantum_complexity + decryption_failure_complexity
    #print("\n")
    print("(k, quantum, classical, dfp)=", k, quantum_complexity, classical_complexity, decryption_failure_complexity)
    return quantum_complexity

# Generate values for k
k_values = np.arange(1, 1.5, 0.05)

# Variant-512
rank = 2 #2,3,4
n = 256
logDFP = -139
logDFP = -146.1
m = 2*rank*n
threshold = sqrt(2/log(e,2) - 2*logDFP/log(e,2))
final_complexities1_q = [compute_complexity_q(k, threshold, m) for k in k_values]

# Variant-768
rank = 3 #2,3,4
n = 256
logDFP = -164
logDFP = -177.8
m = 2*rank*n
threshold = sqrt(2/log(e,2) - 2*logDFP/log(e,2))
final_complexities2_q = [compute_complexity_q(k, threshold, m) for k in k_values]

# Variant-1024
rank = 4 #2,3,4
n = 256
logDFP = -174
logDFP = -167.6
m = 2*rank*n
threshold = sqrt(2/log(e,2) - 2*logDFP/log(e,2))
final_complexities3_q = [compute_complexity_q(k, threshold, m) for k in k_values]

```

```python
for i in range(len(k_values)):
    print(format(k_values[i], '.2f'), end = ", \t")
    print(format(final_complexities1_q[i],'.2f'), end = ", \t")
    print(format(final_complexities2_q[i],'.2f'), end = ", \t")
    print(format(final_complexities3_q[i],'.2f'))
```
