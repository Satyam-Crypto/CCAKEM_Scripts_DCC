# CCAKEM-512

```python
from estimator import *

params = LWE.Parameters(
    n=512,
    q=2048,
    Xs=ND.CenteredBinomial(1),
    Xe=ND.CenteredBinomial(3),
    m=512
)

res = LWE.estimate.rough(params)
print(res)
LWE.arora_gb(params)
```

```python
beta = res["usvp"]["beta"]

print("Classical Core-SVP Hardness = ", 0.292*beta)
print("Quantum Core-SVP Hardness = ", 0.265*beta)
```
